# Project Overview

This project focuses on performing end-to-end SQL-based data analysis on a realistic e-commerce dataset. The complete workflow starts with generating synthetic datasets, followed by data cleaning, database creation, SQL analysis, and Python-SQL integration. The project demonstrates how SQL can be used to extract meaningful business insights from transactional data.

---

# Project Objectives

The main objectives of this project are:

- Generate realistic e-commerce datasets.
- Clean and validate the generated data.
- Store cleaned datasets in a SQLite database.
- Perform SQL-based business analysis.
- Apply Window Functions and Common Table Expressions (CTEs).
- Perform Customer Segmentation and Cohort Analysis.
- Integrate Python with SQLite.
- Understand real-world SQL analysis workflow.

---

# Project Workflow

The project is completed in the following phases:

1. Data Generation
2. Data Cleaning
3. SQLite Database Creation
4. SQL Analysis
5. Python + SQLite Integration

Each phase builds upon the previous phase to complete the overall data analysis workflow.

---

# Data Generation

## Purpose

The first phase of the project is to generate realistic e-commerce datasets using Python.

The Faker library is used to generate realistic customer information such as names and email addresses. Randomly generated transaction records are also created to simulate an actual e-commerce platform.

The following datasets are generated:

- Customers
- Products
- Orders
- Order Items

Intentional errors such as missing values, invalid email addresses, incorrect date formats, and negative quantities are introduced so that the data cleaning process can be demonstrated in the next phase.

---

## Observation

The required datasets were generated successfully.

Generated Files:

- customers.csv
- products.csv
- orders.csv
- order_items.csv

These datasets serve as the raw input for the complete project.

---

# Data Cleaning

## Purpose

The generated datasets intentionally contain several inconsistencies.

The purpose of this phase is to improve the overall quality of the data before loading it into the database.

The following cleaning operations were performed:

- Incorrect date formats were corrected.
- Missing customer IDs were handled.
- Product names were standardized.
- Invalid email addresses were validated and corrected.
- Negative quantities were corrected.
- Referential integrity was verified.
- Data quality report was generated.

---

## Observation

The data cleaning process completed successfully.

All required datasets were cleaned and validated.

The cleaned datasets are stored inside the **data/cleaned** folder and are used throughout the remaining phases of the project.

# SQLite Database Creation

## Purpose

After cleaning the datasets, they are loaded into a SQLite database.

The following tables were created:
