# Week 7 - Delta Lake MERGE Implementation

## Overview

In this assignment, I learned how to perform incremental data processing using Delta Lake. The assignment includes loading data into a Delta table, cleaning the data, creating incremental data, applying the MERGE operation, validating the results, and displaying the final dataset.

## Objective

The objective of this assignment is to understand how Delta Lake performs incremental data processing using the MERGE operation. This assignment also covers data cleaning, updating existing records, inserting new records, and validating the final output.

# Environment Setup

For this assignment, I used Databricks Free Edition with Python and Delta Lake support. The notebook was created in Databricks to perform all the required data processing tasks.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("DeltaLakeAssignment").getOrCreate()

print("Spark Session Created Successfully")
```

# Dataset Preparation

For this assignment, the Superstore dataset was uploaded into Databricks and stored as a Delta table named `customer_master`. This table will be used throughout the assignment for data cleaning, MERGE operations, and validation.

## Step 1 - Load Dataset into a Delta Table
The uploaded Delta table is loaded into a Spark DataFrame. This DataFrame will be used in the remaining steps of the assignment.

```python
# Load the Delta table

customer_df = spark.table("workspace.default.customer_master")

customer_df.show(5)

customer_df.printSchema()
```

### Result

The Delta table was loaded successfully into a Spark DataFrame. The first few records and the schema were displayed.

## Step 2 - Perform Basic Data Cleaning
In this step, the dataset is cleaned by checking for null values and removing duplicate records. This helps improve data quality before performing the MERGE operation.

```python
from pyspark.sql.functions import col, count, when

# Check null values in each column
null_values = customer_df.select([
    count(when(col(column).isNull(), column)).alias(column)
    for column in customer_df.columns
])

null_values.show()
```

```python
# Remove duplicate records

cleaned_df = customer_df.dropDuplicates()

print("Original Row Count:", customer_df.count())
print("Row Count After Removing Duplicates:", cleaned_df.count())
```

### Result

The dataset was checked for duplicate records. No duplicate records were found, so the total number of records remained the same after applying the duplicate removal operation.

## Step 3 - Create Incremental Dataset

In this step, I created a second dataset to simulate incremental data. I updated the **Sales** value for a few existing records and added one new customer record. This dataset will be used in the next step to perform the Delta Lake **MERGE** operation.

```python
from pyspark.sql.functions import col, lit

# Take first 5 records from cleaned dataset
incremental_df = cleaned_df.limit(5)

# Update Sales value to simulate modified records
incremental_df = incremental_df.withColumn(
    "Sales",
    col("Sales") + 100
)

# Create one new record
new_record = cleaned_df.limit(1)

new_record = (
    new_record
    .withColumn("Order ID", lit("NEW-10001"))
    .withColumn("Customer ID", lit("CG-99999"))
    .withColumn("Customer Name", lit("New Customer"))
)

# Combine updated records with new record
incremental_df = incremental_df.unionByName(new_record)

# Display incremental dataset
display(incremental_df)
```

### Result

The incremental dataset was created successfully. It contains both updated records and one new record, which will be merged with the main Delta table in the next step.

## Step 4 - Apply MERGE Operation

In this step, I performed the Delta Lake MERGE operation using the incremental dataset. Existing records were updated based on the matching Order ID, and new records were inserted into the Delta table.

```python
from pyspark.sql.functions import col

# Delta Lake does not support spaces in column names.
# Therefore, the columns are renamed by replacing spaces with underscores
# before saving the DataFrame as a Delta table.

cleaned_delta_df = cleaned_df.toDF(
    *[column.replace(" ", "_") for column in cleaned_df.columns]
)

incremental_delta_df = incremental_df.toDF(
    *[column.replace(" ", "_") for column in incremental_df.columns]
)

cleaned_delta_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("customer_delta")

print("Delta table created successfully")
```

### Note

While creating the Delta table, I got an error because some column names contained spaces. To solve this issue, I replaced the spaces with underscores before saving the data as a Delta table.

```python
from delta.tables import DeltaTable

delta_table = DeltaTable.forName(spark, "customer_delta")

(
    delta_table.alias("target")
    .merge(
        incremental_delta_df.alias("source"),
        "target.Order_ID = source.Order_ID"
    )
    .whenMatchedUpdateAll()
    .whenNotMatchedInsertAll()
    .execute()
)

print("MERGE Operation Completed Successfully")
```

```python
merged_df = spark.table("customer_delta")

display(merged_df)
```

### Result

The MERGE operation was completed successfully. Existing records were updated, and the new record was inserted into the Delta table.

## Step 5 - Validate the Results
After performing the MERGE operation, I validated the final dataset by checking the total number of records and verifying whether any duplicate records were present.

```python
# Load the merged Delta table
merged_df = spark.table("customer_delta")

# Display total number of records
print("Total Rows:", merged_df.count())
```

### Result

The total number of records was displayed successfully after the MERGE operation.

```python
# Check duplicate records using Order_ID

duplicate_count = (
    merged_df.groupBy("Order_ID")
    .count()
    .filter("count > 1")
    .count()
)

print("Duplicate Records:", duplicate_count)
```

### Result

The validation was completed successfully. The total number of records after the MERGE operation is **9995**.

The duplicate count based on **Order ID** is expected because one order can contain multiple products in the Superstore dataset. Therefore, these are valid records and not data quality issues.

# Step 6 - Display Final Dataset and Summary

In the final step, I displayed the updated Delta table after applying the MERGE operation. This helps verify that the existing records were updated and the new record was inserted successfully.


```python
# Load final Delta table
final_df = spark.table("customer_delta")

# Display first 10 records
final_df.show(10, truncate=False)
```

```python
# Display total number of records
print("Final Total Records:", final_df.count())
```

### Result

The final Delta table was displayed successfully after performing the MERGE operation.

The dataset now contains both the updated records and the newly inserted record. The final row count confirms that the incremental data was processed successfully using the Delta Lake MERGE operation.
